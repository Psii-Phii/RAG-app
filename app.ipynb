```python
#Splitting the data
from langchain.text_splitter import RecursiveCharacterTextSplitter, CharacterTextSplitter, TokenTextSplitter

```

```python
#load pdf

loader = PyPDFLoader("/Users/ritikkotak/Desktop/LC/Carroll_1898_Alice_in_Wonderland.pdf")
pages = loader.load()
text_splitter = CharacterTextSplitter(
    separator= "\n", 
    chunk_size = 1000, 
    chunk_overlap = 500,
    length_function = len
)
```

```python
docs = text_splitter.split_documents(pages)
print(docs[30])
print(len(pages))
print(len(docs))
```

```python
#Splitting based on tokens 
token_text_splitter = TokenTextSplitter( 
    chunk_size = 100, 
    chunk_overlap = 50)

token_docs = token_text_splitter.split_documents(pages)

print(token_docs[75]) #adds metadata for better context 
```

```python
#Vector store/ Embeddings

#what if we wanted to work with multiple PDFs

loaders = [
    PyPDFLoader("/Users/ritikkotak/Desktop/LC/Carroll_1898_Alice_in_Wonderland.pdf"),
    PyPDFLoader("/Users/ritikkotak/Desktop/LC/Through_the_Looking_glass_and_what_Alice.pdf")
]

docs = []

for loader in loaders:
    docs.extend(loader.load())
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 250)
splits = text_splitter.split_documents(docs)
len(splits)
```

```python
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores import Chroma
```

```python
persist_directory  = "/Users/ritikkotak/Desktop/LC/Chroma/"
```

```python
! rm -rf /Users/ritikkotak/Desktop/LC/Chroma/
```

```python
vectordb = Chroma.from_documents(
    documents=splits,
    embedding=OpenAIEmbeddings(),
    persist_directory= persist_directory
)
```

```python
print(vectordb._collection.count())


```

```python
#similarity search

question = "what kind of rabbit did Alice see? Describe it in detail"
docs = vectordb.similarity_search(question, k=5)
print(docs[0].page_content)
```

```python
for doc in docs:
    print(doc.metadata)
```

```python
#We still need to improve upon our retrival 
#MMR And LLM aided retirval 

mmr_docs = vectordb.max_marginal_relevance_search(question, k=2, fetch_k= 5)
mmr_docs[0].page_content[0:500]
```

```python
mmr_docs[1].page_content[0:500]
```

```python
from langchain.llms import OpenAI
from langchain.retrievers.self_query.base import SelfQueryRetriever
from langchain.chains.query_constructor.base import AttributeInfo

for d in mmr_docs:
    print(d.metadata)
```

```python
metadata_field_info = [
    AttributeInfo(name = "source", description="This excerpt is from the book , should be one of Alice in Wonderland/ Through the looking glaa",
                   type= "string",
                   ),
    AttributeInfo(name="page", description="The page from the book", type="integer",),
]
```

```python

document_content_description = "Novel"
llm = OpenAI(temperature = 0)
retriever = SelfQueryRetriever.from_llm(llm,vectordb, document_content_description,metadata_field_info,verbose= True)
```

```python
docs = retriever.get_relevant_documents(question)
```
